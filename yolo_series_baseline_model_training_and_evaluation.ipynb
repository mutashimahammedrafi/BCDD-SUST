**Model Training**

```python
from ultralytics import YOLO
import datetime
import torch

# Log environment details for the paper
print(f"CUDA available: {torch.cuda.is_available()}")
print(f"GPU: {torch.cuda.get_device_name(0)}")

model = YOLO('yolo11x.pt')

results = model.train(
    data='data.yaml',
    epochs=100,
    optimizer='SGD',
    imgsz=640,
    batch=32,
    patience=10,

)
```

**Model Evaluation**

```python
from ultralytics import YOLO
import pandas as pd

model = YOLO("best.pt")


results = model.val(data="data.yaml", split="test")

class_indices = results.box.ap_class_index
class_names = [results.names[idx] for idx in class_indices]

precision = results.box.p
recall = results.box.r
f1_scores = results.box.f1
map50 = results.box.ap50
map50_95 = results.box.ap

classwise_metrics = pd.DataFrame({
    "Class Name": class_names,
    "Precision": precision,
    "Recall": recall,
    "F1-Score": f1_scores,
    "mAP50": map50,
    "mAP50-95": map50_95
})

print("\n" + "="*50)
print("CLASS-WISE TEST METRICS")
print("="*50)
print(classwise_metrics.to_string(index=False))
```
